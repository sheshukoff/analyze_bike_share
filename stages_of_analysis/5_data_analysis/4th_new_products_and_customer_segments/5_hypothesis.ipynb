- **Гипотеза 5.** На сколько можно увеличить количество аренды велосипедов сдавая их как походные или как выезд на природу (от 1 до нескольких дней).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

- читаю файл "../../../data_after_preprocessing/drd_with_weather.parquet"

```python
drd_with_weather = pd.read_parquet('../../../data_after_preprocessing/drd_with_weather.parquet')
drd_with_weather
```

- выбираю поездки, которые длятся более суток (это более 1440 минут)

```python
long_rent_bike = drd_with_weather[drd_with_weather['duration_in_minutes'] >= 1440]
long_rent_bike
```

- решаю отсортировать данные по столбцу started_at

```python
long_rent_bike_sorted = long_rent_bike.sort_values(by='started_at')
```

- делаю группировку по столбцу rideable_type и member_casual и считаю количество поездок

```python
long_rent_bike_sorted.groupby(['rideable_type', 'member_casual'])['duration_in_minutes'].count()
```

- хочу посмотреть только классически байки

```python
long_rent_bike_sorted = long_rent_bike_sorted[long_rent_bike_sorted['rideable_type'] == 'classic_bike']
long_rent_bike_sorted
```

- заменяю nan на 0

```python
long_rent_bike_sorted['end_station_id'] = long_rent_bike_sorted['end_station_id'].replace(np.nan, 0)
```

- вывожу описание для столбца end_station_id <> 0

```python
long_rent_bike_sorted[long_rent_bike_sorted['end_station_id'] != 0]['duration_in_minutes'].describe()
```

- делаю группировку предварительно оставив только те станции куда известно поставили велосипед

```python
long_rent_bike_sorted[long_rent_bike_sorted['end_station_id'] != 0].groupby(['rideable_type', 'member_casual'])['duration_in_minutes'].count()
```
