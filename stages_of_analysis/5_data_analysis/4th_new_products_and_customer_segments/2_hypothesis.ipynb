- **Гипотеза 2.** Можно ли увеличить количество аренды велосипедов (любых) сдавая их компании из нескольких человек (4 или более) или корпоративным клиентам.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

- читаю файл "../../../data_after_preprocessing/drd_with_weather.parquet"

```python
drd_with_weather = pd.read_parquet('../../../data_after_preprocessing/drd_with_weather.parquet')
drd_with_weather
```

- убираю станции, которые равны 0

```python
drd_with_weather = drd_with_weather[(drd_with_weather['start_station_id'] != 0) & (drd_with_weather['end_station_id'] != 0)]
```

- сортирую данные по столбцам 'start_station_id' ,'started_at'

```python
sorted_ride_a_bike = drd_with_weather.sort_values(by=['start_station_id' ,'started_at'])
```

- создаю новую переменную part_df

```python
part_df = sorted_ride_a_bike
```

- оставляю столбцы 'started_at', 'ended_at', 'start_station_id', 'end_station_id'

```python
part_df = part_df[['started_at', 'ended_at', 'start_station_id', 'end_station_id']]
part_df
```

- удаляю отсутствующие значения

```python
part_df = part_df.dropna()
```

- вывожу результат

```python
part_df
```

- смотрю сколько осталось станций 'start_station_id'

```python
list_start_station_id = part_df['start_station_id'].unique()
len(list_start_station_id)
```

- смотрю сколько осталось станций 'end_station_id'

```python
list_end_station_id = part_df['end_station_id'].unique()
len(list_end_station_id)
```

- создаю переменную dict_for_index

```python
dict_for_index = {}
```

- написал скрипт, который ищет совместные поездки

```python
for start_station_id in list_start_station_id:
    print(start_station_id)
    # здесь использую только часть df на 100к строк
    df = part_df[part_df['start_station_id'] == start_station_id]
    for end_station_id in list_end_station_id:
        together_ride = df[df['end_station_id'] == end_station_id]
        if len(together_ride) >=4: 
            # код ниже нужно сделать функцию которая вернет dataframe со всеми группами
            # print("end_station_id", len(together_ride), end_station_id)
            diff_to_down = together_ride['started_at'].diff().dt.total_seconds().div(60).reset_index()
            diff_to_down['started_at'] = diff_to_down['started_at'].replace({np.nan: 0})
            diff_to_down['bool_series'] =  diff_to_down['started_at'] > 3
            diff_to_down['groups'] = diff_to_down['bool_series'].cumsum()
            # print(diff_to_down)
            # код ниже нужно сделать функцию которая вернет индексы для поиска в исходнике
            group_df = diff_to_down.groupby('groups')['index'].count()
            group_more_four = group_df[group_df >= 4]
            
            if len(group_more_four.values) != 0:
                dict_for_index[f'group_{start_station_id}_to_{end_station_id }'] = list(group_more_four.values)
                # print(len(group_more_four.values), 'len(group_more_four.values)')
                # print(group_more_four.values, 'group_more_four.values')
                # код ниже нужно сделать функцию
                idexes = diff_to_down[diff_to_down['groups'].isin(group_more_four.index)]
                dict_for_index[f'station_{start_station_id}_to_{end_station_id}'] = idexes['index']
```

- вывожу результат

```python
dict_for_index
```

- считаю сколько групп получилось

```python
count = 0
for station, group in dict_for_index.items():
    if 'group_' in station:
        count += len(group)
```

- вывожу результат

```python
count
```

- создаю переменную list_indexes, что бы затем найти эти индексы в исходных данных

```python
list_indexes = []
```

- записываю индексы в список list_indexes

```python
for station, indexes in dict_for_index.items():
    if 'station_' in station:
        for index in indexes.values:
            list_indexes.append(index)
```

- вывожу результат (это количество строк)

```python
len(list_indexes)
```

- создаю DataFrame, куда запишу индексы

```python
result = pd.DataFrame(
    {
        'index': list_indexes
    }
)

result
```

- сбрасываю индексы в DataFrame drd_with_weather, что бы затем сделать merge с переменной result

```python
drd_with_weather = drd_with_weather.reset_index()
```

- вывожу результат

```python
drd_with_weather
```

- записываю данные в excel

```python
result.merge(drd_with_weather, how='inner', on='index').to_excel('together_ride_full.xlsx', sheet_name='together_ride', index=False)
```

- сейчас буду проверять как завершаются совместные поездки, читаю файл excel, который только что записал

```python
together_ride = pd.read_excel('together_ride_full.xlsx', sheet_name='together_ride')
together_ride
```

- вывел info

```python
together_ride.info()
```

- добавляю столбец change_time, что бы потом отбросить раницу во времени более 3 минут

```python
together_ride['change_time'] = together_ride['ended_at'].diff().dt.total_seconds().div(60)
```

- зменяю значения Nan на 0

```python
together_ride['change_time'] = together_ride['change_time'].replace({np.nan: 0})
```

- делаю числа по модулю

```python
together_ride['change_time'] = together_ride['change_time'].abs()
```

- здесь обрасываю разницу во времени более 3 минут

```python
together_ride['bool_series'] = together_ride['change_time'] > 3
```

- добавляю новый столбец groups, что бы посчитать количество групп совместных поездок

```python
together_ride['groups'] = together_ride['bool_series'].cumsum()
```

- вывожу результат

```python
together_ride
```

- делаю группировку по столбцу groups и считаю сколько входит в эту группу аренд велосипеда

```python
temp_df_2 = together_ride.groupby('groups')['index'].count()
temp_df_2
```

- выбираю только те группы, которые состовляют 4 и более совместных аренд

```python
index_groups = temp_df_2[temp_df_2 >= 4]
```

- вывожу результат

```python
index_groups
```
